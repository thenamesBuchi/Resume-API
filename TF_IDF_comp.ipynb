```python
import spacy
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Download the SpaCy model if not already present
!python -m spacy download en_core_web_md

# Load SpaCy
nlp = spacy.load("en_core_web_md")

resume = "I am a student skilled in Python and SQL databases"
job_desc = "Looking for a developer with experience in Python and PostgreSQL"

# 1. TF-IDF
tfidf_vec = TfidfVectorizer()
tfidf_matrix = tfidf_vec.fit_transform([resume, job_desc])
tfidf_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])[0][0]

# 2. SpaCy (Word2Vec)
doc1 = nlp(resume)
doc2 = nlp(job_desc)
spacy_sim = doc1.similarity(doc2)

print(f"TF-IDF Similarity: {tfidf_sim:.4f}")
print(f"SpaCy (Word2Vec) Similarity: {spacy_sim:.4f}")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import spacy
from sentence_transformers import SentenceTransformer, util

# 1. Load Data
df = pd.read_csv('/content/resume_dataset_1200.csv')

# 2. Select a Sample (Let's use Akash Pillai at Index 0)
resume_text = df.iloc[0]['Skills']
job_desc_1 = df.iloc[0]['Target_Job_Description']
job_desc_2 = df.iloc[1]['Target_Job_Description']

# --- MODEL 1: TF-IDF ---
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform([resume_text, job_desc_1, job_desc_2])
tfidf_scores = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:])
print(f"TF-IDF Scores: Job 1: {tfidf_scores[0][0]:.4f}, Job 2: {tfidf_scores[0][1]:.4f}")

# --- MODEL 2: SpaCy (Word2Vec) ---
#
nlp = spacy.load("en_core_web_md")
doc_res = nlp(resume_text)
score_spacy_1 = doc_res.similarity(nlp(job_desc_1))
score_spacy_2 = doc_res.similarity(nlp(job_desc_2))
print(f"SpaCy Scores: Job 1: {score_spacy_1:.4f}, Job 2: {score_spacy_2:.4f}")

# --- MODEL 3: Sentence-BERT ---
sbert_model = SentenceTransformer('all-MiniLM-L6-v2')
emb_res = sbert_model.encode(resume_text)
emb_j1 = sbert_model.encode(job_desc_1)
emb_j2 = sbert_model.encode(job_desc_2)
score_sbert_1 = util.cos_sim(emb_res, emb_j1)
score_sbert_2 = util.cos_sim(emb_res, emb_j2)
print(f"SBERT Scores: Job 1: {score_sbert_1.item():.4f}, Job 2: {score_sbert_2.item():.4f}")
```
